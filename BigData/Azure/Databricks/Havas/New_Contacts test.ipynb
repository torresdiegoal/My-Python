```python
##############  ENTRANCES  #################

# Diligenciar las dos variables de snapshots. Una vez diligenciado, ejecutar este comando con ctrl + Enter

snapshot_to_analyse = '20250131' # snapshot que indica el mes que se quiere estudiar
snapshot_prev = '20241231' # snapshot inmediatamente anterior a snapshot_to_analyse


# -----------------
date_snapshot_to_analyse = f"{snapshot_to_analyse[:4]}-{snapshot_to_analyse[4:6]}-{snapshot_to_analyse[6:]}" # snapshot_to_analyse en formato aaaa-mm-dd
date_snapshot_prev = f"{snapshot_prev[:4]}-{snapshot_prev[4:6]}-{snapshot_prev[6:]}" # snapshot_prev en formato aaaa-mm-dd

date_snapshot_to_analyse_01 = f"{snapshot_to_analyse[:4]}-{snapshot_to_analyse[4:6]}-01" # snapshot_prev en formato aaaa-mm-01

query = f"""
with analysis as (
  -------------------
  select
    brand_country,
    brand_code,
    count(distinct case when snapshot_date_key = '{snapshot_to_analyse}' then brand_customer_id end) 
      - count(distinct case when snapshot_date_key = '{snapshot_prev}' then brand_customer_id end)  as substraction_between_snaps,
    0 as lifetime_new_consumer_flag,
    0 as first_interaction_in_month
  from crm_reporting.fact_segment_by_brand
  group by all

  union all

  --------------------
  select
    brand_country,
    brand_code,
    0 as substraction_between_snaps,
    count(distinct case when lifetime_new_consumer_activity_segment_key = 'DDM_LNCAS_1C' then brand_customer_id end) as lifetime_new_consumer_flag,
    0 as first_interaction_in_month
  from crm_reporting.fact_segment_by_brand
  where snapshot_date_key in ('{snapshot_to_analyse}')
  group by all

  union all

  -------------------
  select 
    brand_country,
    brand_code,
    0 as substraction_between_snaps,
    0 as lifetime_new_consumer_flag,
    count(distinct source_customer_id) as first_interaction_in_month
  from (
      select distinct
        brand_country,
        brand_code,
        source_customer_id,
        FIRST_VALUE(date(interaction_dt)) 
          OVER (PARTITION BY brand_country,brand_code,source_customer_id ORDER BY interaction_dt asc) AS interaction_dt
      from prod_latam_catalog.crm_reporting.fact_customer_interactions
  ) 
  where interaction_dt between '{date_snapshot_to_analyse_01}' and '{date_snapshot_to_analyse}'
  group by all 
  --------------------
)

select
  brand_country,
  brand_code,
  sum(substraction_between_snaps) as substraction_between_snaps,
  sum(lifetime_new_consumer_flag) as lifetime_new_consumer_flag,
  sum(first_interaction_in_month) as first_interaction_in_month,
  round(sum(lifetime_new_consumer_flag) / sum(first_interaction_in_month),2) as new_contacts_diff_rate -- the closest to 1, the better
from analysis
group by all
having sum(substraction_between_snaps) > 0 
  --and sum(lifetime_new_consumer_flag) > 0
  --and sum(first_interaction_in_month) > 0
order by brand_country,brand_code
"""
result_df = spark.sql(query)
display(result_df)

```
