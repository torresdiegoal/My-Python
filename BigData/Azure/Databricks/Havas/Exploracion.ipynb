```python
# Import packages and functions needed 
from pyspark.sql import functions as F

### Merged tables
merge = spark.sql("""      
select * 
from
  (select distinct
  a.brand_country,
  a.brand_code,
  c.is_deleted,
  -- a.is_deleted as is_deleted_dc,
  -- a.delete_reason,
  concat(year(a.created_dt),'_',month(a.created_dt)) as year_month,
  a.source_customer_id as source_customer_id_dc,
  b.source_customer_id as source_customer_id_dcb
  from prod_latam_catalog.crm_reporting.dim_customer a
  inner join prod_latam_catalog.crm_reporting.dim_customer_extn c
  on a.source_customer_id = c.source_customer_id
  and a.brand_code = c.brand_code
  and a.brand_country = c.brand_country
  left join prod_latam_catalog.crm_reporting.dim_customer_bridge b
  on a.source_customer_id = b.source_customer_id
  and a.brand_code = b.brand_code
  and a.brand_country = b.brand_country
)
where is_deleted is null
  and year_month like ('2024_%')
order by year_month

""")

# ids not found in bridge table
# tmp = merge.filter(F.col("source_customer_id_dcb").isNull())
# display(tmp)


merge.createOrReplaceTempView("merge_vw")

tmp = spark.sql("""
select
  year_month,
  --count(distinct source_customer_id_dc) as num_ids_dim_customer,
  count(distinct case when source_customer_id_dcb is null then source_customer_id_dc end) as ids_missing_in_bridge--,
  -- count(distinct case when source_customer_id_dcb is not null then source_customer_id_dc end) as no_nulos
from merge_vw
group by all
order by year_month
""")

display(tmp)
# print(merged_gdm.count()) #140280
```

```python
%sql
select * 
from
  (select distinct
  a.brand_country,
  a.brand_code,
  c.is_deleted,
  concat(year(a.created_dt),'_',month(a.created_dt)) as year_month,
  a.source_customer_id as source_customer_id_dc,
  b.source_customer_id as source_customer_id_dcb
  from prod_latam_catalog.crm_reporting.dim_customer a
  inner join prod_latam_catalog.crm_reporting.dim_customer_extn c
  on a.source_customer_id = c.source_customer_id
  and a.brand_code = c.brand_code
  and a.brand_country = c.brand_country
  left join prod_latam_catalog.crm_reporting.dim_customer_bridge b
  on a.source_customer_id = b.source_customer_id
  and a.brand_code = b.brand_code
  and a.brand_country = b.brand_country
  where upper(a.source_name) IN ('DEMANDWARE','JEBBIT','SITECORE') 
    AND (lower(a.registration_source) LIKE ('%quiz%') 
      OR lower(a.registration_source) LIKE ('%diagnos%')
      OR lower(a.registration_source) LIKE ('%website%')))
where source_customer_id_dcb is null
and is_deleted is null
order by year_month
```

```python
%sql
with tmp as (
  select
    concat(year(a.created_dt),'_',month(a.created_dt)) as year_month,
    a.source_customer_id,
    c.is_deleted,
    a.is_deleted as is_deleted_dc
  from prod_latam_catalog.crm_reporting.dim_customer a
  left join prod_latam_catalog.crm_reporting.dim_customer_extn c
    on a.source_customer_id = c.source_customer_id
    and a.brand_code = c.brand_code
    and a.brand_country = c.brand_country
    and a.source_name = c.source_name
  where year(a.created_dt) = 2024
    and c.is_deleted is null
)

select
  year_month,
  count(case when is_deleted == 'Y' then source_customer_id end) as is_deleted_extn,
  count(case when is_deleted_dc == 'Y' then source_customer_id end) as is_deleted_dc
from tmp
group by year_month
order by year_month
```
