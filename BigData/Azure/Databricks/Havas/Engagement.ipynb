#### D2C EXAMPLE

```python
%sql
-- D2C BRANDS
select distinct 
  fci.interaction_type,
  fci.interaction_sub_type 
from (
  select * 
  from crm_reporting.fact_customer_interactions) fci
  left join (
    select interaction_type,interaction_sub_type,source_name
    from crm_reporting.dim_excl_interaction_type
    where UPPER(interaction_source) = 'KNOWN'
    group by interaction_type,interaction_sub_type,source_name ) dft 
  on UPPER(fci.interaction_type) = UPPER(dft.interaction_type)
  and UPPER(fci.interaction_sub_type) = UPPER(dft.interaction_sub_type)
  where dft.interaction_type IS NULL 
    and dft.interaction_sub_type IS NULL 
    and dft.source_name IS NULL
```

```python
%sql
-- NON D2C ANAlYSIS LAST 12 MONTHS
select 
  -- b.d2c_flag,
  -- a.brand_country,
  -- a.brand_code,  
  a.interaction_sub_type,
  count(distinct c.brand_customer_id) as brand_customer_id
  -- sum(c.brand_customer_id) as 12_months_consumer_activity_segment_cnt
from prod_latam_catalog.crm_reporting.fact_customer_interactions a
-- left join prod_latam_catalog.crm_reporting.dim_brand b
--   on a.brand_code = b.brand_code
--   and a.brand_country = b.brand_country
left join prod_latam_catalog.crm_reporting.etl_stg_customer_bridge c
  on a.brand_code = c.brand_code
  and a.brand_country = c.brand_country
  and a.source_customer_id = c.source_customer_id
------- QUERY PARA OBTENER LAS INTERACCIONES USADAS -------
inner join (
  select distinct 
    fci.interaction_type,
    fci.interaction_sub_type 
  from (
    select * 
    from crm_reporting.fact_customer_interactions) fci
    left join (
      select interaction_type,interaction_sub_type,source_name
      from crm_reporting.dim_excl_interaction_type
      where UPPER(interaction_source) = 'KNOWN'
      group by interaction_type,interaction_sub_type,source_name ) dft 
    on UPPER(fci.interaction_type) = UPPER(dft.interaction_type)
    and UPPER(fci.interaction_sub_type) = UPPER(dft.interaction_sub_type)
    where dft.interaction_type IS NULL 
      and dft.interaction_sub_type IS NULL 
      and dft.source_name IS NULL) d
----------------------------------------------------------
  on a.interaction_type = d.interaction_type
  and a.interaction_sub_type = d.interaction_sub_type
where a.interaction_dt between '2025-02-01' and '2025-02-28'
-- where a.interaction_dt between '2024-02-01' and '2025-01-31'
  -- and a.interaction_sub_type = 'Profile Created'
  -- and a.interaction_sub_type in ('Open','Click')
  and a.brand_country = 'ARG'
  and a.brand_code = 'LAN'
  --and b.d2c_flag = 'Y'
group by all
-- order by brand_country,brand_code
order by interaction_sub_type
```
