```python
import pandas as pd
```

```python
tab_1 = pd.read_excel('excel.xlsx', sheet_name  = 'Nuevos')
print(tab_1.head())
# Eliminar espacios en blanco al principio y al final de la columna VEHICLE_NAME
tab_1['VEHICLE_NAME'] = tab_1['VEHICLE_NAME'].str.strip()

print(' ')

tab_2 = pd.read_excel('excel.xlsx', sheet_name  = 'Completo')
print(tab_2.head())
# Eliminar espacios en blanco al principio y al final de la columna VEHICLE_NAME
tab_2['Nombre'] = tab_2['Nombre'].str.strip()
```

```python
tb = pd.merge(tab_1, tab_2, how='left', left_on='VEHICLE_NAME', right_on='Nombre')

#print(tb)
```

```python
tb.to_excel('resultado.xlsx', index=False)
```
