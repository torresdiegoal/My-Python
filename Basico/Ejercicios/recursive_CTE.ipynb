RECURSIVE CTE'S (COMMON TABLE EXPRESSION )

```python
empleados = [
    {"id": 1, "nombre": "Juan", "id_jefe": None},
    {"id": 2, "nombre": "Ana", "id_jefe": 1},
    {"id": 3, "nombre": "Luis", "id_jefe": 2},
    {"id": 4, "nombre": "Pedro", "id_jefe": 2},
    {"id": 5, "nombre": "María", "id_jefe": 3},
]

# Encontrar todos los empleados bajo Juan (id=1)
def encontrar_subordinados(empleados, jefe_id):
    resultados = []
    stack = [jefe_id]  # Usamos una pila para recorrer la jerarquía

    while stack:
        actual = stack.pop() #.pop() en listas → Elimina y devuelve el último elemento (o uno por índice).
                             #.pop() en diccionarios → Elimina y devuelve el valor de una clave.
        for emp in empleados:
            if emp["id_jefe"] == actual:
                resultados.append(emp)
                stack.append(emp["id"])  # Agregamos al stack para seguir buscando

    return resultados

print(encontrar_subordinados(empleados, 1))
```
